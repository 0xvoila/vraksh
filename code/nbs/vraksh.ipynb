# About English Language

Every word in English language belongs to one category of English Language. Lets call it `word type` 
A word could be one of the following type

1. Noun
2. Pronoun
3. Verb
   1. transistive verb - must have an object
   2. in-transistive verb - do not need object like I sleep 
   3. Linking Verb - in state of like am, is, were, become 
   4. auxillary verb - have, has, was etc
   5. modal verb - must, should, could
4. Adjective
5. Adverb
6. Prepositions
7. Interjection
8. Conjugation

Along with the type of the word, some types may enhance the meaning of another word. For example 
1. Word of type `adjective` may enhance the type of `noun` word. 
2. Word of type `adverb` may enchance the type of the word `verb`. 
3. Word of type `verb` may relates two or more `nouns` 
4. Word of type `pronoun` defines word of type `noun` 
5. Word of type `preposition` defines the location / time of the `verb` or `noun`

````python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_trf")
````

````python
# Simple sentence with nsub, root and dobject
doc = nlp("I go to office regularly at 9:30 AM Morning")
displacy.render(doc, style="dep", jupyter=True)

for token in doc:
    print( f"{token.text} - {token.ent_type_} - {token.pos_}")
````

````python
# Simple sentence with nsub, root and dobject and advmod (slowly)
doc = nlp("Amit eats fruits slowly")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and dobject and advcl (whole clause act as modifier defines why, what of main verb)
doc = nlp("Amit eats fruits while he travels")
displacy.render(doc, style="dep", jupyter=True)

doc = nlp("Amit eats fruits if they are sweat")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and acomp ( adjective complement when main verb is linking verb ) 
doc = nlp("Amit feels good")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and amod ( adjective modifier when adjective link to a noun ) 
doc = nlp("Amit feels good boy")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and xcomp ( open clausal complement ) 
doc = nlp("Amit is going to play cricket")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and ccomp ( clipped clausal complement ) 
doc = nlp("Amit eats fruits and drink milk")

for token in doc:
    print(token, token.ent_type_)

displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and ccomp ( clipped clausal complement ) and preposition psub and pobj
doc = nlp("Amit said that he will play with ball")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and ccomp ( clipped clausal complement ) and preposition psub and pobj
doc = nlp("Amit said that he will go to school")
displacy.render(doc, style="dep", jupyter=True)
````

````python
# Simple sentence with nsub, root and cc
doc = nlp("Amit play with bat and ball")
displacy.render(doc, style="dep", jupyter=True)
````
